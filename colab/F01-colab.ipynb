## 1. 环境与内置资源准备
首次运行安装独立依赖，后续核对安装标记后复用。保留pip报告和包版本；不把安装成功当科学验收。

准备单元已默认隐藏长代码。点击左侧运行按钮即可，不需要阅读或修改内置数据。若手动展开看到很长的字母数字串，那是压缩课程资源的Base64编码，不是乱码，也不是需要学习的化学分析代码；完成后可切回表单视图。

```python
# @title 准备课程环境（点击左侧运行，无需阅读内置数据）
LESSON = 'F01'
BUNDLE_SHA256 = 'd0c8f53be209777006ca636fdb97417cdc27ae9cf30adcf35318d0565af937df'
BUNDLE_B64 = ''
"""Template appended after embedded BUNDLE_B64 and BUNDLE_SHA256 constants."""
import base64, hashlib, io, json, os, platform, subprocess, sys, uuid, venv, zipfile
from pathlib import Path, PurePosixPath

if sys.version_info < (3,12):
    raise RuntimeError('本课要求Python>=3.12；请检查Colab运行时。不能跳过检查。')
IN_COLAB = 'google.colab' in sys.modules
try:
    import google.colab
    IN_COLAB = True
except ImportError:
    pass
if not IN_COLAB and not os.environ.get('ADP_COLAB_PREFLIGHT_ROOT'):
    raise RuntimeError('本文件为Colab适配版；本地预检需由项目验收脚本显式指定隔离目录。')
base=Path(os.environ.get('ADP_COLAB_PREFLIGHT_ROOT',str(Path.cwd()/'adp-p1-colab'))).resolve()
base.mkdir(parents=True,exist_ok=True)
data=base64.b64decode(BUNDLE_B64,validate=True)
if hashlib.sha256(data).hexdigest()!=BUNDLE_SHA256:raise ValueError('内置资源包哈希错误')
run_root=base/(LESSON+'-'+uuid.uuid4().hex);run_root.mkdir()
with zipfile.ZipFile(io.BytesIO(data)) as z:
    if sum(i.file_size for i in z.infolist())>32*1024*1024:raise ValueError('资源超预算')
    for info in z.infolist():
        rel=PurePosixPath(info.filename)
        if rel.is_absolute() or '..' in rel.parts or ':' in info.filename or '\\' in info.filename or (info.external_attr>>16)&0o170000==0o120000:raise ValueError('非法资源路径')
    z.extractall(run_root)
for item in json.loads((run_root/'payload-manifest.json').read_text('utf-8')):
    rel=PurePosixPath(item['file'])
    if rel.is_absolute() or '..' in rel.parts or ':' in item['file'] or '\\' in item['file']:raise ValueError('非法清单路径')
    if hashlib.sha256((run_root/item['file']).read_bytes()).hexdigest()!=item['sha256']:raise ValueError('课程资源哈希错误')
private_env=base/'course-env-r2';python=private_env/('Scripts/python.exe' if os.name=='nt' else 'bin/python')
signature=hashlib.sha256((run_root/'requirements-colab.txt').read_bytes()).hexdigest()
marker=private_env/'adp-installed.json'
if marker.exists() and json.loads(marker.read_text('utf-8')).get('requirements_sha256')!=signature:
    raise RuntimeError('已有环境与课程要求不同，请选择新的工作目录；不覆盖原环境。')
if not marker.exists():
    if private_env.exists():raise RuntimeError('存在未完成安装，请保留日志并选择新的工作目录，不自动删除。')
    venv.EnvBuilder(with_pip=False).create(private_env)
    command=[sys.executable,'-m','pip','--python',str(python),'--isolated','install','--disable-pip-version-check','--no-cache-dir']
    wheels=os.environ.get('ADP_COLAB_PREFLIGHT_WHEELS')
    if wheels:command+=['--no-index','--find-links',wheels]
    else:command+=['--index-url','https://pypi.org/simple']
    command+=['-r',str(run_root/'requirements-colab.txt'),'--report',str(base/'pip-report.json')]
    installed=subprocess.run(command,capture_output=True,text=True,timeout=600)
    (base/'pip-install.log').write_text(installed.stdout+'\n'+installed.stderr,encoding='utf-8')
    if installed.returncode:raise RuntimeError('课程依赖安装失败；查看adp-p1-colab/pip-install.log，不把失败当通过。')
    wheel=next((run_root/'wheels').glob('*.whl'))
    subprocess.run([sys.executable,'-m','pip','--python',str(python),'--isolated','install','--no-index','--no-deps',str(wheel)],check=True,capture_output=True,timeout=120)
    check=subprocess.run([sys.executable,'-m','pip','--python',str(python),'check'],capture_output=True,text=True,timeout=60)
    if check.returncode:raise RuntimeError('私有环境pip check失败：'+check.stdout)
    freeze=subprocess.check_output([sys.executable,'-m','pip','--python',str(python),'freeze','--all'],text=True,timeout=60)
    # Keep package versions only; pip may otherwise include the absolute wheel source.
    versions=[]
    for line in freeze.splitlines():
        if ' @ ' not in line:versions.append(line)
    versions.append('ai4orgchem-data-processing==0.1.0')
    (base/'resolved-environment.json').write_text(json.dumps({'packages':sorted(versions),'python':platform.python_version(),'platform':platform.system()},indent=2),encoding='utf-8')
    marker.write_text(json.dumps({'requirements_sha256':signature,'pip_check':'PASS','environment':'private venv; Colab base environment unchanged','execution_location':'COLAB' if IN_COLAB else 'LOCAL_PREFLIGHT_NOT_COLAB'},indent=2),encoding='utf-8')
for src,dest in [(marker,'installation.json'),(base/'pip-install.log','pip-install.log'),(base/'resolved-environment.json','resolved-environment.json'),(base/'pip-report.json','pip-report.json')]:
    if src.exists():(run_root/dest).write_bytes(src.read_bytes())
print('资源完整性PASS；独立环境安装/复用完成。下一单元设置参数。Colab预装依赖未被覆盖。')
```

## 3. 从头运行原课
每次独立新内核，保留全部断言、拒绝例和预期FAIL。完整教学输出随后显示。

```python
import shutil, time
os.environ['PYTHONIOENCODING']='utf-8'
os.environ['PYTHONUTF8']='1'
parameter = COURSE_PARAMETER
if type(parameter) is not bool:raise TypeError('参数类型不符')
if parameter not in [False,True]:raise ValueError('首轮适配仅接受默认值和声明的练习值')
experiment=run_root/('experiment-'+uuid.uuid4().hex)
experiment.mkdir()
for path in run_root.iterdir():
    if path.name.startswith('experiment-'):continue
    if path.is_dir() and path.name in ['samples','wheels']:shutil.copytree(path,experiment/path.name)
    elif path.is_file():shutil.copy2(path,experiment/path.name)
(experiment/'parameters.json').write_text(json.dumps({'lesson':LESSON,'key':'OMIT_LAST_TERM','value':parameter},ensure_ascii=False,indent=2),encoding='utf-8')
executed=subprocess.run([str(python),'-I','-B',str(experiment/'runner.py'),str(experiment),str(experiment/'parameters.json')],capture_output=True,text=True,encoding='utf-8',timeout=600)
(experiment/'kernel-execution.log').write_text(executed.stdout+'\n'+executed.stderr,encoding='utf-8')
if executed.returncode:raise RuntimeError('课程执行失败；查看本次kernel-execution.log，不能报告通过。')
receipt=json.loads((experiment/'run-receipt.json').read_text('utf-8'))
print(json.dumps(receipt,ensure_ascii=False,indent=2))
```

## 4. 查看整课输出并独立解释
显示原课程段落、真实执行文字、表格和图片；核心断言通过不评价解释能力。F01漏项应FAIL，F02全MO仍DIAGNOSTIC。

```python
from IPython.display import display, Markdown, HTML, Image
completed=json.loads((experiment/'executed.ipynb').read_text('utf-8'))
for cell in completed['cells']:
    if cell['cell_type']=='markdown':display(Markdown(''.join(cell['source'])))
    else:
        for output in cell.get('outputs',[]):
            if output['output_type']=='stream':print(''.join(output['text']))
            data=output.get('data',{})
            if 'image/png' in data:display(Image(data=base64.b64decode(''.join(data['image/png']))))
            elif 'text/html' in data:display(HTML(''.join(data['text/html'])))
            elif 'text/plain' in data:print(''.join(data['text/plain']))
archive=experiment/'colab-results.zip'
with zipfile.ZipFile(archive) as z:
    assert z.testzip() is None
    for item in json.loads(z.read('export-manifest.json')):
        assert hashlib.sha256(z.read(item['file'])).hexdigest()==item['sha256']
print('保存/重载/ZIP清单校验PASS；文件：colab-results.zip；SHA-256：'+hashlib.sha256(archive.read_bytes()).hexdigest())
```
